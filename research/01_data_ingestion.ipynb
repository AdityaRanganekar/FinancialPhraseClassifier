```python
import os
from pathlib import Path
from dataclasses import dataclass
from datasets import load_dataset
from box import ConfigBox
import yaml
```

```python
os.chdir("../")
```

```python
with open("config/config.yaml") as f:
    content = ConfigBox(yaml.safe_load(f))
```

```python
@dataclass(frozen=True)
class DataIngestionConfig:
    root_dir: Path
    dataset_name: str
    local_data_file: Path

config = content.data_ingestion
data_ingestion_config = DataIngestionConfig(
    root_dir=Path(config.root_dir),
    dataset_name=config.dataset_name,
    local_data_file=Path(config.local_data_file)
)
```

```python
os.makedirs(data_ingestion_config.root_dir, exist_ok=True)
dataset = load_dataset(data_ingestion_config.dataset_name)
dataset.save_to_disk(data_ingestion_config.local_data_file)
print("Data ingestion prototype verified:", os.listdir(data_ingestion_config.local_data_file))
```
